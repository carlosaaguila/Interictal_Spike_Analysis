```python
import pandas as pd
import ast
import re

# Read the CSV file
df = pd.read_csv('HUP_files_v2.csv', index_col=0)

def process_channel_labels(channels):
    # Convert string representation of list to actual list
    channel_list = ast.literal_eval(channels)
    
    def filter_channel(channel):
        # Use regex to separate alphabetic prefix from numeric suffix
        match = re.match(r'([A-Za-z]+)(\d+)', channel)
        if match:
            prefix, number = match.groups()
            # Keep the channel if it's not exactly 'LH' or 'RH'
            return prefix not in ['LH', 'RH']
        # Keep channels that don't match the pattern (e.g., 'LAST1')
        return True
    
    # Filter the channels
    filtered_channels = [ch for ch in channel_list if filter_channel(ch)]
    
    return filtered_channels

# Apply the processing function to the channel_label column
df['channel_label'] = df['channel_label'].apply(process_channel_labels)

# Display the first few rows of the processed dataframe
print(df[['filename', 'pt_id', 'channel_label', 'processed_channels']].head())

# Save the processed dataframe to a new CSV file
df.to_csv('processed_HUP_files.csv')

print("\nProcessed data saved to 'processed_HUP_files.csv'")
```
